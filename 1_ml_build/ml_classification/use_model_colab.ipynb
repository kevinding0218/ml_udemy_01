```python
classifier_colab = pickle.load(open("classifier.pickle", "rb"))
sc_colab         = pickle.load(open("sc.pickle", "rb"))

new_input = np.array([[40, 20000]])  # [age, salary] — same column order as training

prediction  = classifier_colab.predict(sc_colab.transform(new_input))
probability = classifier_colab.predict_proba(sc_colab.transform(new_input))[:, 1]

print(f"Prediction : {prediction[0]}  (1 = bought, 0 = didn't buy)")
print(f"Confidence : {probability[0]:.0%}")
```
